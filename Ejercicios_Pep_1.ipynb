```
Problema 1
La conjetura de Collatz es un problema aún no resuelto en matemáticas que
postula que a partir de cualquier número entero positivo n es posible
llegar al valor 1 con dos reglas sencillas:
    Si n es par, se divide por dos.
    Si n es impar, se multiplica por tres y se le suma 1.

Según esta conjetura, si se repite este proceso un número finito de veces,
sin importar cuál sea el número inicial, se llegará al resultado final
de 1. Actualmente la conjetura de Collatz está comprobada para todos los
valores enteros menores a 560, por lo que se solicita a usted que
construya un programa en Python que muestre el camino que se realiza para
probar la conjetura con un entero n. Considere que la entrada es un
número entero positivo.

Casos de ejemplo:
    Entrada 1: 5
    Salida 1: 5 16 8 4 2 1
    Entrada 2: 7
    Salida 2: 7 22 34 17 52 26 13 40 20 10 5 16 8 4 2 1

```

````python
# NO HAY CONSTANTES

 # ENTRADA
 #Se piede el número para comprobar la conjetura de Collatz
numero = int(input("Ingrese un número entero positivo: "))

 # PROCESAMIENTO
#Mientras el numero sea diferente a 1 se continua el ciclo
while numero != 1:
    #Se imprime el numero
    print(numero, end=" ")
    #Si el numero es par se divide entre 2
    if numero % 2 == 0:
        numero = numero // 2
    #Si el numero es impar se multiplica por 3 y se le suma 1
    else:
        numero = numero * 3 + 1

# SALIDA
#Se imprime el último
print(numero)
````

```
Problema 2
Un panagrama es una oración que contiene todas las letras del alfabeto,
por ejemplo, en español, la oración:
    * “Whisky bueno: ¡excitad mi pequeña y frágil vejez!”
Es un panagrama, por otro lado, la oración:
    * “Por un momento estuve en un extraño sillón de felpa chirriante”
No es un panagrama, pues no están presentes las letras b, g, h, j, k, q,
w, x, y, z.
Construya un programa en Python que reciba como entrada un string e
informe al usuario si este es un panagrama o no. Considere para su
implementación que:
    * El texto recibido puede venir con mayúsculas y minúsculas.
    * El programa recibirá sólo oraciones en inglés, como por ejemplo “The quick brown fox jumps over the lazy dog”,
    por lo que ni los tildes, ni la letra ñ deben considerarse.
```

````python
# CONSTANTES
ALFABETO = "abcdefghijklmnopqrstuvwxyz"

# ENTRADA
frase = input("Ingrese una oración: ").lower()

# PROCESAMIENTO
es_panagrama = True
for caracter in ALFABETO:
    encontrado = False

    for letra in frase:
        if caracter == letra:
            encontrado = True

    if not encontrado:
        es_panagrama = False

# SALIDA
if es_panagrama:
    print("La oración es un panagrama.")
else:
    print("La oración no es un panagrama.")
````

```
Problema 4
Matías es un ornitólogo muy meticuloso que se encuentra actualmente ordenando
la base de datos de aves de un zoológico, para ello tiene un registro del
nombre científico del animal y el peso promedio de este en gramos. Como
usualmente le solicitan que entregue la información ordenada por tamaño,
desde el ave más pequeña a la más grande, Matías desea un programa en el
que pueda agregar ambos datos de cada ave y se vaya generando automáticamente
una lista ordenada, añadiendo cada ave en la posición que le corresponda
por tamaño. En caso de que Matías ingrese un ave que ya ha ingresado, el
programa debería informárselo.
Para verificar la factibilidad de la solución propuesta por Matías, se le
ha solicitado a usted escribir un programa en Python que reciba pares de
la forma [nombreCientifico, peso] y los agregue a una lista ordenada por
tamaño. Considere que, en caso de haber dos aves con el mismo peso, se
deberá añadir el ave nueva a continuación de la existente.
Un ejemplo del funcionamiento del programa para la lista listaAves dada
sería:
    * listaAves = [[aaa, 23.4], [abc, 32.5], [cdc,1440.3], [vgr, 7000.0]]
    Se prohíbe utilizar métodos y funciones nativas para insertar en una posición determinada,
    para ordenar la lista o para obtener máximos y mínimos.
```

````python
# NO HAY CONSTANTES

# ENTRADA Y PROCESAMIENTO
lista_inicial = []
programa_cerrado = False
while programa_cerrado != True:
    print("\n--- Menú de Gestión de Aves ---")
    opcion = input("Ingrese una opción:\n1. Añadir ave y peso\n2. Cerrar programa\nOpción: ")

    if opcion == "1":
        ave = input("Ingrese el nombre científico del ave: ")
        ya_existe = False
        for ave_existente in lista_inicial:
            if ave == ave_existente[0]:
                ya_existe = True

        if ya_existe:
            print("Ya ingresó el ave.")

        else:
            peso = float(input("Ingrese el peso del ave: "))
            lista_inicial.append([ave, peso])

    elif opcion == "2":
        print("\nPrograma cerrado.")
        programa_cerrado = True

for i in range(len(lista_inicial) - 1):
    indice_minimo = i
    for j in range(i + 1, len(lista_inicial)):
        if lista_inicial[j][1] < lista_inicial[indice_minimo][1]:
            indice_minimo = j

    temp = lista_inicial[i]
    lista_inicial[i] = lista_inicial[indice_minimo]
    lista_inicial[indice_minimo] = temp

# SALIDA
print("Los resultados son:", lista_inicial)
````
